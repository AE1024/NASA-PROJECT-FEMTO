```python
import os

os.makedirs("C:\\Users\\elmaz\\AIProject\\NASA_Project\\raw_data", exist_ok=True)
```

```python
from huggingface_hub import hf_hub_download

files = ["features_train.parquet", "features_val.parquet", "features_test.parquet"]

for fname in files:
    path = hf_hub_download(
        repo_id="Amgharr/FEMTO-ST_DATASET",
        filename=fname,
        repo_type="dataset",
        local_dir="C:\\Users\\elmaz\\AIProject\\NASA_Project\\raw_data",
    )
    print(f" {fname} -> {path}")
```

```python
#check if the files are downloaded
for f in files:
    if os.path.exists(f"C:\\Users\\elmaz\\AIProject\\NASA_Project\\raw_data\\{f}"):
        print(f"{f} exists")
    else:
        print(f"{f} does not exist")
```

```python
#check the size of the files
for f in files:
    if os.path.exists(f"C:\\Users\\elmaz\\AIProject\\NASA_Project\\raw_data\\{f}"):
        size = os.path.getsize(f"C:\\Users\\elmaz\\AIProject\\NASA_Project\\raw_data\\{f}")
        size_mb = size / (1024 * 1024)
        print(f"{f} size: {size_mb:.2f} MB")
        
```
